```python
import random
import string
import tkinter as tk
from tkinter import simpledialog, messagebox
import pyperclip  # Biblioteca para copiar texto para a área de transferência

def gerar_senha(tamanho=12, usar_maiusculas=True, usar_numeros=True, usar_simbolos=True, ordem_aleatoria=True):
    """
    Gera uma senha aleatória de acordo com os critérios especificados.
    
    :param tamanho: O tamanho da senha (padrão: 12 caracteres).
    :param usar_maiusculas: Incluir letras maiúsculas (padrão: True).
    :param usar_numeros: Incluir números (padrão: True).
    :param usar_simbolos: Incluir símbolos especiais (padrão: True).
    :param ordem_aleatoria: Se True, embaralha os caracteres (padrão: True).
    :return: Uma senha.
    """
    letras = string.ascii_lowercase
    if usar_maiusculas:
        letras += string.ascii_uppercase
    
    numeros = string.digits if usar_numeros else ""
    simbolos = string.punctuation if usar_simbolos else ""
    
    senha = []
    
    # Garante que pelo menos um caractere de cada tipo esteja presente
    if usar_maiusculas or letras:
        senha.append(random.choice(letras))
    if usar_numeros:
        senha.append(random.choice(numeros))
    if usar_simbolos:
        senha.append(random.choice(simbolos))
    
    # Adiciona o restante dos caracteres
    restantes = tamanho - len(senha)
    if ordem_aleatoria:
        # Junta tudo e escolhe aleatoriamente
        todos_caracteres = letras + numeros + simbolos
        senha += random.choices(todos_caracteres, k=restantes)
        random.shuffle(senha)
    else:
        # Respeita a ordem (Letra, Número, Caractere)
        while restantes > 0:
            if letras and restantes > 0:
                senha.append(random.choice(letras))
                restantes -= 1
            if numeros and restantes > 0:
                senha.append(random.choice(numeros))
                restantes -= 1
            if simbolos and restantes > 0:
                senha.append(random.choice(simbolos))
                restantes -= 1
    
    return ''.join(senha)

# Inicializa a interface gráfica
root = tk.Tk()
root.withdraw()  # Oculta a janela principal do Tkinter

# Coletando as opções do usuário
try:
    tamanho_senha = int(simpledialog.askstring("Tamanho da Senha", "Digite o tamanho da senha desejada (ex: 12):"))
    usar_maiusculas = messagebox.askyesno("Letras Maiúsculas", "Incluir letras maiúsculas?")
    usar_numeros = messagebox.askyesno("Números", "Incluir números?")
    usar_simbolos = messagebox.askyesno("Símbolos", "Incluir símbolos?")
    ordem_aleatoria = messagebox.askyesno("Ordem Aleatória", "Gerar em ordem aleatória?")

    # Gerar a senha
    senha = gerar_senha(tamanho_senha, usar_maiusculas, usar_numeros, usar_simbolos, ordem_aleatoria)

    # Copiar a senha para a área de transferência e exibir a mensagem
    pyperclip.copy(senha)  # Copia a senha para a área de transferência
    messagebox.showinfo("Senha Gerada", f"Sua senha gerada é:\n{senha}\n\nA senha foi copiada para a área de transferência!")

except ValueError:
    messagebox.showerror("Erro", "Você não digitou um número válido para o tamanho da senha!")
```
